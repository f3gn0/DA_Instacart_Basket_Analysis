```python
# Step 5 - Run a check for missing values in your df_ords data-frame
df_ords.isnull().sum()
```

```python
# There are a total of 206209 missing values in the column days_since_prior_order. 
# An explanation for this quanitty of missing values could be customers' first orders don't have a value.
# However, we must keep this information instead of remove it.
```

```python
# Step 6 - Address the missing values using an appropriate method.
```

```python
# To determine what course of action to take regarding the missing values, we must separate the data from the original dataframe,
df_ords_nan = df_ords[df_ords['days_since_prior_order'].isnull() == True]
```

```python
df_ords_nan
```

```python
# Now we must verify whether all 'order_number' values equal to 1 are missing values in the 'days_since_prior_order' column
```

```python
# Filter the dataframe where 'order_number' equals 1\n
df_first_orders = df_ords_nan[df_ords_nan['order_number'] == '1']
```

```python
# Check if all of these have missing values in 'days_since_prior_order'
df_first_orders_missing = df_first_orders['days_since_prior_order'].isna().all()
```

```python
# Print the result
print("All first orders have missing 'days_since_prior_order':", df_first_orders_missing)
```

```python
# We know for certain that every row with 'order_number' equal to 1 has a missing value in the'days_since_prior_order' column
```

```python
# Address the missing values by creating a column that flags order numbers as either True/False depending on if the 'order_number' = 1\n",
df_ords['first_order'] = df_ords['days_since_prior_order'].isna()
```
