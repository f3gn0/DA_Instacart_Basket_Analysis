```python
# Step 5 - Determine a suitable way to combine the orders_products_combined dataframe with your products data set
```

```python
# The column "product_id" is a key identifier shared between these two data-frames, which makes a merge operation feasible.
# The type of merge to apply depends on the desired content of the final dataframe.
# Since we aim to retain only meaningful information that exists in both dataframes, an 'inner' merge is the most suitable option.
# This ensures that only rows with matching product_id values in both dataframes are included in the final result.
```

```python
# Check columns in df_ords_prod_comb
print('Columns in df_ords_prod_comb:', df_ords_prod_comb.columns.tolist())
```

```python
# Check columns in df_prods
print('Columns in df_prods:', df_prods.columns.tolist())
```

```python
# The dataframe df_ords_prod_comb has already a column named "_merge" which conflicts with the default name of the indicator column.
# When using "indicator=True" pandas tries to create a new column called _merge, but it cannot overwrite an existing column with the
# same name. Consequently we need to use a custom name for the function.
```

```python
# test merge without overwriting with custom name the for new merge column
pd.merge(
    df_ords_prod_comb,
    df_prods,
    on=['product_id'],
    how='inner',
    indicator='merge_info'  # Rename the indicator column
)
```

```python
# The test was ok.
# The column _merge created from the previous exercise is of no use for the new data-frame df_ords_prod_comb.
# We should remove the column _merge after the new merge is completed.
```

```python
# Merge data-frame df_prods and orders_products_combined
ords_prods_merge = df_ords_prod_comb.merge(
    df_prods, 
    on=['product_id'], 
    how='inner', 
    indicator='merge_info'
)
```

```python
# Drop the column "_merge" from the previous merge
ords_prods_merge.drop('_merge', axis=1, inplace=True)
```

```python
# Check after dropping the column
print(ords_prods_merge.columns)
```

```python
# Drop the column "'Unnamed: 0_x'" from the previous exercise
ords_prods_merge.drop('Unnamed: 0_x', axis=1, inplace=True)
```

```python
# Check after dropping the column
print(ords_prods_merge.columns)
```
