```python
# Define a function to sort products data-frame ords_prods_merge according to price
def price_label(row):
  if row['prices'] <= 5:
    return 'Low-range product'
  elif (row['prices'] > 5) and (row['prices'] <= 15):
    return 'Mid-range product'
  elif row['prices'] > 15:
    return 'High range'
  else: return 'Not enough data'
```

```python
# Apply the function
df_subset['price_range']=df_subset.apply(price_label, axis=1)
```

```python
# Step 2 #2 Suppose your clients have changed their minds about the labels you created in
# your “busiest_day” column. Now, they want “Busiest day” to become “Busiest days” (plural).
# This label should correspond with the two busiest days of the week as opposed to the single busiest day.
# At the same time, they’d also like to know the two slowest days. Create a new column for this using a suitable method.
```

```python
result2 = []
for value in ords_prods_merge["order_dow"]:
  if value == 0 or value == 1: 
    result2.append("Busiest days")
  elif value == 3 or value == 4:
    result2.append("Lowest Busy days")    
  else:
    result2.append("Regularly busy days")
```

```python
#  Check the Length of result2
print(len(result2))
```

```python
# Combining the result with the data-fram to create a new column
ords_prods_merge['Busiest_days']=result2
```

```python
# Step 4 - identify the busiest hours of the day. Rather than by hour, they want periods of time labeled
# “Most orders,” “Average orders,” and “Fewest orders.”
# Create a new column containing these labels called “busiest_period_of_day"
```

```python
# create list of hours sorted by value count
hours_sorted=ords_prods_merge['order_hour_of_day'].value_counts(dropna=False).index.to_numpy()
```

```python
# Create an empty list and use a for-loop to populate it with busiest time of day labels
busiest_hour = [] # Initialize an empty list
for row in ords_prods_merge['order_hour_of_day']:  # Iterate over the column
    if row in hours_sorted[:8]:  # Check if the hour is in the first 8 hours
        busiest_hour.append("Most orders")  # Append "Most orders"
    elif row in hours_sorted[8:16]:  # Check if the hour is in the next 8 hours
        busiest_hour.append("Average orders")  # Append "Average orders"
    else:  # For all other hours
        busiest_hour.append("Fewest orders")  # Append "Fewest orders"
```

```python
# create column in dataframe equal to busiest_hour
ords_prods_merge['busiest_period_of_day'] = busiest_hour
```
