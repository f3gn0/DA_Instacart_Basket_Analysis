Step 6 - Create a spending flag for each user based on the average price across all their orders using the following criteria:
If the mean of the prices of products purchased by a user is lower than 10, then flag them as a “Low spender.”
If the mean of the prices of products purchased by a user is higher than or equal to 10, then flag them as a “High spender.”

```python
# Calculating average spend per user
ords_prods_merge['average_spend'] = ords_prods_merge.groupby('user_id')['prices'].transform('mean')
```

```python
# Creating the flag for Low_spender customers
ords_prods_merge.loc[ords_prods_merge['average_spend'] < 10, 'spending_flag'] = 'Low_spender'
```

```python
# Creating the flag for High_spend customers
ords_prods_merge.loc[ords_prods_merge['average_spend'] >= 10, 'spending_flag'] = 'High_spender'
```

```python
# Check customers spender flags
ords_prods_merge['spending_flag'].value_counts(dropna = False)
```

Step 7 - Create an order frequency flag that marks the regularity of a user’s ordering behavior according to the median in the days_since_prior_order” column.
The criteria for the flag should be as follows:
If the median of “days_since_prior_order” is higher than 20, then the customer should be labeled a “Non-frequent customer.”
If the median is higher than 10 and lower than or equal to 20, then the customer should be labeled a “Regular customer.”
If the median is lower than or equal to 10, then the customer should be labeled a “Frequent customer.”

```python
# Create a column for frequent customers/users
ords_prods_merge['Customer_frequency'] = ords_prods_merge.groupby(['user_id'])['days_since_prior_order'].transform('median')
```

```python
# Creating the flag for condition 1 (Non-frequent customer)
ords_prods_merge.loc[ords_prods_merge['Customer_frequency'] > 20, 'frequency_flag'] = 'Non-frequent customer'
```

```python
# Creating the flag for condition 2 (Regular customer)
ords_prods_merge.loc[(ords_prods_merge['Customer_frequency'] > 10) & (ords_prods_merge['Customer_frequency'] <= 20), 'frequency_flag'] = 'Regular customer'
```

```python
# creating the flag for condition 3 (Frequent customer)
ords_prods_merge.loc[ords_prods_merge['Customer_frequency'] <= 10, 'frequency_flag'] = 'Frequent customer'
```

```python
# Check customers frequency flag
ords_prods_merge['frequency_flag'].value_counts(dropna = False)
```

Steps to Verify the hypothesis related to NaN values:
Count the Number of Unique Orders per User:

Extract users whose frequency_flag is NaN and check if they match those with only one order.

```python
# Count the number of unique orders for each user
user_order_counts = ords_prods_merge.groupby('user_id')['order_number'].nunique()
```

```python
# Identify users with only one order
single_order_users = user_order_counts[user_order_counts == 1].index
```

```python
# Check if these users correspond to the NaN values in the frequency_flag column
nan_users = ords_prods_merge[ords_prods_merge['frequency_flag'].isna()]['user_id'].unique()
```

```python
# Compare the two sets of users
single_order_nan_match = set(single_order_users) == set(nan_users)
```

```python
# Output the result
print("Do NaN values correspond to single-order users?", single_order_nan_match)
print("Number of users with NaN frequency_flag:", len(nan_users))
print("Number of users with only one order:", len(single_order_users))
```

The results indicate that the NaN values in the frequency_flag column do not correspond exclusively to users with only one order. Specifically:
