## Working with Python Data Types

```python
a = 50
b = 50
result = a + b
print("Result:", result)  # Output: 100
```

```python
x = 150
y = 50
result = x - y
print("Result:", result)  # Output: 100
```

```python
p = 70
q = 40
r = 10
result = p + q - r
print("Result:", result)  # Output: 100
```

```python
# Define two floating-point variables
num1 = 25.5
num2 = 5.1

# Perform division
result = num1 / num2

# Print the result
print("Result:", result)  # Output: 5.0
```

```python
# Define separate strings
part1 = "Py"
part2 = "th"
part3 = "on"

# Combine the strings to form a word
word = part1 + part2 + part3

# Print the result
print("Constructed word:", word)  # Output: Python
```

```python
# Define separate strings for the first sentence
part1 = "I"
part2 = " love"
part3 = " Python."

# Combine the strings to form the first sentence
sentence1 = part1 + part2 + part3

# Define separate strings for the second sentence
part4 = "Coding"
part5 = " is"
part6 = " fun!"

# Combine the strings to form the second sentence
sentence2 = part4 + part5 + part6

# Print the sentences
print("First sentence:", sentence1)  # Output: I love Python.
print("Second sentence:", sentence2)  # Output: Coding is fun!
```
